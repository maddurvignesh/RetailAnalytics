```python
numeric = df.select_dtypes(include=np.number)

for col in numeric.columns:
    plt.figure(figsize=(6,4))
    plt.hist(df[col].dropna(), bins=20, edgecolor='black')
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
categorical = df.select_dtypes(include='object')

for col in categorical.columns:
    plt.figure(figsize=(6,4))
    df[col].value_counts().plot(kind='bar')
    plt.title(f"Bar Plot of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
for col in categorical.columns:
    plt.figure(figsize=(6,6))
    df[col].value_counts().plot(kind='pie', autopct='%1.1f%%')
    plt.title(f"Pie Chart of {col}")
    plt.ylabel("")
    plt.show()
```

```python
for col in numeric.columns:
    plt.figure(figsize=(6,4))
    plt.boxplot(df[col].dropna())
    plt.title(f"Outliers in {col}")
    plt.ylabel(col)
    plt.show()
```

```python
for col in numeric.columns:
    plt.figure(figsize=(6,4))
    plt.scatter(range(len(df)), df[col])
    plt.title(f"Scatter Plot of {col}")
    plt.xlabel("Index")
    plt.ylabel(col)
    plt.show()
```

```python
## Task 4: Bivariate Analysis
```

```python
import matplotlib.pyplot as plt
import numpy as np

corr = df.corr(numeric_only=True)

plt.figure(figsize=(14,10))
plt.imshow(corr, cmap='coolwarm', aspect='auto')
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")
plt.show()
```

```python
numerical_cols = df.select_dtypes(include=np.number).columns

for col in numerical_cols:
    if col != "Response":
        plt.figure(figsize=(6,4))
        plt.scatter(df[col], df["Response"])
        plt.xlabel(col)
        plt.ylabel("Response")
        plt.title(f"{col} vs Response")
        plt.show()
```

```python
categorical = df.select_dtypes(include='object').columns

for col in categorical:
    pd.crosstab(df[col], df["Response"]).plot(kind='bar', figsize=(7,4))
    plt.title(f"{col} vs Response")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
from scipy.stats import chi2_contingency

for col in categorical:
    table = pd.crosstab(df[col], df["Response"])
    chi2, p, dof, expected = chi2_contingency(table)

    print(f"{col}")
    print(f"Chi-square Statistic = {chi2:.2f}")
    print(f"P-value = {p:.4f}")
    print("-"*40)
```

```python
for col in numerical_cols:
    if col != "Response":
        plt.figure(figsize=(6,4))
        df.boxplot(column=col, by="Response")
        plt.title(f"{col} by Response")
        plt.suptitle("")
        plt.show()
```

Analysis Report: Exploratory Data Analysis (EDA) for Retail Analytics Fraud Detection

Exploratory Data Analysis (EDA) was performed using Python to understand the characteristics of the retail customer dataset and identify patterns that could support fraud detection and customer behavior analysis. The dataset was first examined by checking its structure, data types, and missing values. This helped ensure that the data was suitable for further analysis and highlighted the need to handle missing values, particularly in the Income column.

Descriptive statistical analysis was carried out to calculate measures such as mean, median, minimum, maximum, and standard deviation for all numerical variables. These statistics provided insights into customer demographics, income levels, purchasing behavior, and spending across different product categories. Histograms and boxplots were used to visualize the distribution of numerical variables and identify outliers. Several spending-related features, such as Wine, Meat Products, and Gold Products, contained high-value outliers that may represent premium customers or potentially unusual purchasing behavior.

Univariate analysis was performed using histograms for numerical variables and bar charts and pie charts for categorical variables such as Education and Marital Status. This analysis revealed the distribution of customers across different demographic groups and highlighted the most common customer profiles. Boxplots further confirmed the presence of extreme values in spending variables, which are important for identifying anomalies.

Bivariate analysis was conducted to examine relationships between customer attributes and the marketing campaign response. Correlation analysis showed positive relationships among several spending categories, indicating that customers who spend more on one product category often spend more on others. Scatter plots and boxplots suggested that customers with higher incomes generally exhibited higher spending patterns. Bar plots and chi-square tests indicated that categorical variables such as education and marital status may influence customer responses to marketing campaigns.

From a fraud detection perspective, EDA helps identify abnormal customer behavior, unusual purchase amounts, and unexpected transaction patterns that may indicate fraudulent activities. Customers with spending patterns significantly different from the majority can be flagged for further investigation. The insights obtained from EDA also support feature selection, improve data quality, and provide a strong foundation for developing machine learning models capable of detecting fraudulent transactions and predicting customer responses.

Overall, the EDA process provided valuable insights into customer behavior, purchasing trends, and potential anomalies, enabling data-driven decision-making and supporting the development of effective retail analytics and fraud detection systems.
